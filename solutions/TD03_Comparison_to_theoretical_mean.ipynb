## 2. Bilateral test 


**Objective:**

**We would like to know whether the Hamilton scores observed in Group 1 women
women are equal on average to 29.**.

We need to perform a Student's t test comparing the mean estimated by the function
function with the expected theoretical mean E = 29 in our case.

*The principle of this test is simple*: it checks whether the difference between the observed mean $\mu$ in the sample
theoretical mean $\mu _0$ is sufficiently far from 0 with a certain risk of error
risk of error $\alpha$ for the two averages to be considered different.
The test therefore calculates the difference between $\mu$ and $\mu _0$ and divides by its standard deviation to normalize:

$$ T = \frac{\mu - \mu _0}{\sqrt{\frac{S ^2}{n} }}=\frac{mu - \mu _0}{\frac{S}{\sqrt{n}}}$$

H0: Hamilton score for women in group 1 is equal to 29.

H1: Hamilton score for women in group 1 is not equal to 29.

**Sample extraction**:
- extract the actual sample
- get back its size using the `length()` function

```r
# Data extraction
w1 = depression[depression$sex_g1 == "F", "hamilton_g1"]

# Display extracted data
w1

# Sample size
length(w1)
```

**Checking application conditions**

Check sample size: 
- If the sample size is large ($n \geq 30$): no application condition
- If sample size is small ($n < 30$): data must follow a normal distribution

Check application conditions

Hypotheses of the Shapiro test

H0 : the Hamilton scores of women in group I follow a normal distribution

H1 : the Hamilton scores of women in group I don't follow a normal distribution

```r
# Shapiro test
shapiro.test(w1)
```

**Conclusion :**

We observe p = 0.61 >>> alpha = 0.05, thus we cannot reject H0, so we conclude that the scores of women in group I follow a normal distribution

**Student's t-test**
You can perform the test using the `t.test()` function which takes the sample as a first argument, and the theoretical mean `m` of value `X` as another argument `m = X` (Note: `X` here should be an actual number when calling the function).

```r
# Student's t-test
t.test(w1, m = 29)
```

**Conclusion** :

We observe p = 2.4⋅10⁻⁷ <<< alpha = 0.05, thus we reject H0, which means that the average Hamilton score of women in group I is different from 29.

## 3. One-sided test

**Objective:**

We would like to know whether the Hamilton scores observed in Group 1 women
women are higher on average than 29.


In this case, we will perform a one-tailed test:

H0: the Hamilton score for women in group 1 is equal to 29.

H1: the Hamilton score for women in group 1 is greater than 29.

```r
# One-sided t-test
t.test(w1, m = 29, alternative = "greater")
```

**Conclusion**:

We observe p = 1.2⋅10⁻⁷ <<< alpha = 0.05, thus we reject H0, which means that the average Hamilton score of women in group I is greater than 29.
