### 3.4 Calculating the confidence interval

The mean $m$ calculated from our sample is an estimate of the mean $\mu$ Hamilton score for the female population.
the mean $\mu$ of the Hamilton score for the female population. $m$ fluctuates as a function of the
the sample we are analyzing. It is therefore a random variable with a normal distribution
a normal distribution with expectation $E[m]$ and variance $V[m]$ if Hamilton's score itself is distributed
itself according to a normal distribution with mean $\mu$ and variance $\sigma ^2$.

For a sample of size $n$ and for a confidence interval at a given threshold $\alpha$ (often 95%) is calculated with the following formula:

$$CI = \mu \pm Z_{1-\frac{\alpha}{2}} \frac{\sigma}{\sqrt{n}}$$

with $Z_{1-\frac{\alpha}{2}}$ the quantile of the Student's law.


There is no function in **R** that can be used to calculate a confidence interval directly.
So to calculate the 95% confidence interval of the Hamilton score, we'll use the `CI()` function, which allows you to do this in a single line.

This function will take as arguments the data (x) and the chosen threshold.  
For example, for a 96% confidence interval the function call should be: `CI(x, 0.96)` (4% $\alpha$ threshold).

**IMPORTANT NOTE: that you'll need to copy-paste this function at the top of each new session notebook - an example will be provided in third session notebook**.

```r
# You don't have to understand the code below nor to edit it
# Simply to know how to use it and copy-paste it
C.I <- function(x, alpha){
    alpha <- 1-alpha
    mu = mean(x)
    sigma <- sd(x)
    n = length(x)
    CI_inf = mu - (qt(1 - alpha/2,n-1) * sigma/sqrt(n))
    CI_sup = mu + (qt(1 - alpha/2,n-1) * sigma/sqrt(n))
    CI = c(inf = CI_inf, sup = CI_sup) 
    return(CI)
    
}
```

```r
# Default 95% confidence interval for mean Hamilton score
# of women in group 1
C.I(women_g1, 0.95)
```

The true mean lies between the bounds: $[41.570; 41.703]$

2. Compute the confidence interval of the Hamilton score for men, then for women, for group 2, verifying these conditions of application (at the 95% threshold).

```r
## 1.
# Extract data
men_g1 = depression[depression$sex_g1 == "H", "hamilton_g1"]

# Check sample size
length(men_g1)

# -> Sample size >30 so no need to check normal distribution
# Compute the confidence interval
C.I(men_g1, 0.95)

# For women of group I the conditions of application are checked above
C.I(women_g1, 0.95)

## 2.
men_g2 = depression[depression$sex_g2 == "H", "hamilton_g2"]

# Sample size
length(men_g2)
# => sample size >30 so no need to check normal distribution
C.I(men_g2, 0.95)

women_g2 = depression[depression$sex_g2 == "F", "hamilton_g2"]
length(women_g2)
# => sample size <30 so need to check if normal distribution

## First method: Shapiro Test
# let's perform a shapiro-test
# Hypotheses:
# H0: the Hamilton scores of women in group II follows a normal distribution
# H1: They don't follow a normal distribution
shapiro.test(women_g2)

# We observe p = 0.83 >>> alpha = 0.05, so we can't reject H0, and we conclude that the scores of women in group II follow a normal distribution

## Second method: quantile-quantile plot
qqnorm(women_g2)
qqline(women_g2)
# It seems that the points follow a normal distribution, though with only n = 5, it's difficult to have a clear conclusion anyway

# We can proceed to compute the confidence interval
C.I(women_g2, 0.95)
```
