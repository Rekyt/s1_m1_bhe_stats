## 1. Non-parametric comparison test of several independent samples (Kruskal-Wallis)

### 1.1 Test principle 

The Kruskal-Wallis test is the generalization to K populations of the two-tailed Mann-Whitney test. The data of all samples are replaced by their rank: 

*Example*: A factory wants to compare three different settings on its machine to produce pills, one standard setting and two alternative settings. The measured variable is the number of pills produced per unit of time.
We observe the following results:

|       Setting | Number of Pills | Rank |
|---------------|-----------------|------|
|      Standard |             340 |    5 |
|      Standard |             345 |    9 |
|      Standard |             330 |    1 |
|      Standard |             342 |    6 |
|      Standard |             338 |    3 |
| Alternative 1 |             339 |    4 |
| Alternative 1 |             333 |    2 |
| Alternative 1 |             344 |    8 |
| Alternative 2 |             347 |   10 |
| Alternative 2 |             343 |    7 |
| Alternative 2 |             349 |   11 |
| Alternative 2 |             355 |   12 |

Then we compute 

- $\bar{r}$ the overall mean rank (mean rank for all samples combined)
- $\bar{r_k}$ the average rank for the observations in sample $k$.
- $n$ the cumulative size of all samples      

The principle of this test can be seen simply as follows: 

- if there is no difference between the samples, the distribution of ranks between the samples should be random, which would imply that the samples have approximately the same mean rank, i.e. $\bar{r}$.
- which means that $\bar{r_k}-\bar{r}$ should be close to 0 for each sample.
- in conclusion, if we sum these differences ($\bar{r_k}-\bar{r}$) for all samples, this sum should be close to 0.


### 1.2 Example: Comparing the efficacy of 4 treatments

Different fever blister treatments, including a placebo (A), were randomly assigned to
20 patients. The data give the number of days to complete healing for each treatment.

The data are contained in the file `TD06_fever_blister.txt` in the `data/` folder.

**Question: Are the healing times different according to the treatments used?**

1. Read the data

```r
# Load the data
fever = read.table("data/TD06_fever_blisters.txt", header = TRUE)

# Show the first few lines
head(fever)
```

2. Data presentation and description

```r
# Plot a boxplot representing the number of days as a function of treatment
boxplot(days ~ treatment, data = fever)
```

**Boxplot interpretation**

To check whether there is a real significant difference between the samples, we will perform the Kruskal-Wallis test:

3. Hypotheses 

Specify hypotheses H0 and H1

H0 : The number of fever blisters is the same for all treatment

H1 : At least one treatment has a different number of fever blisters

4. Performing the test 

```r
# Kruskal-Wallis test
kruskal.test(days~treatment, data = fever)
```

5. Conclusion of the test: 

We observe p = 0.18 >>> alpha = 0.05, so we can't reject H0 and as such all the treatments have the same outcome.

## 2. Nonparametric paired-sample comparison test (Friedman)

### 2.1 Test principle 

This is indeed a K-means comparison test, but on paired data.

*Example*: we're interested in comparing three methods to weigh tablets. We have a sample of 10 tablets and we weigh each of them using the three methods. This means that we obtain three different weights for each tablet as reported in the table below

| Tablet | Method 1 | Method 2 | Method 3 |
|--------|----------|----------|----------|
|      1 |        9 |        8 |       10 |
|      2 |       13 |       13 |       12 |
|      3 |       12 |       11 |       10 |
|      4 |       10 |        9 |        8 |
|      5 |       14 |       12 |       13 |
|      6 |       11 |        9 |        9 |
|      7 |        9 |        9 |        8 |
|      8 |       12 |       11 |       12 |
|      9 |        9 |        9 |       10 |
|     10 |        7 |        6 |        6 |

Unlike the Kruskal-Wallis test, the ranks are assigned only within each group (here the three methods for each tablet). The corresponding ranks are shown in the additional columns

| Tablet | Method 1 | Method 2 | Method 3 | Rank 1 ($\hat{r_1}$) | Rank 2 ($\hat{r_2}$) | Rank 3 ($\hat{r_3}$) |
|--------|----------|----------|----------|--------|--------|--------|
|      1 |        9 |        8 |       10 |      2 |      1 |      3 | 
|      2 |       13 |       13 |       12 |    2.5 |    2.5 |      2 |
|      3 |       12 |       11 |       10 |      3 |      2 |      1 |
|      4 |       10 |        9 |        8 |      3 |      2 |      1 |
|      5 |       14 |       12 |       13 |      3 |      1 |      2 |
|      6 |       11 |        9 |        9 |      3 |    1.5 |    1.5 |
|      7 |        9 |        9 |        8 |    2.5 |    2.5 |      1 |
|      8 |       12 |       11 |       12 |    2.5 |      1 |    2.5 |
|      9 |        9 |        9 |       10 |    1.5 |    1.5 |      3 |
|     10 |        7 |        6 |        6 |      3 |    1.5 |    1.5 |

So, we'll be working with ranks, but calculated within blocks.
the Friedman test from all tests for independent samples, where ranks were always
were always based on the total number of observations measured.
If we reason by block, the principle is the same as that of Kruskal-Wallis:

- if there is no difference between samples, the distribution of ranks per block between samples should be random
- which implies that the samples should have approximately the same mean rank, which should correspond to the overall mean rank
- which implies that the differences $\bar{r_k}-\bar{r}$ should be close to 0
- so if we sum these differences for all samples, it should be close to 0.

### 2.2 Example: Does treatment of a herniated disc influence pain perception?

It may be interesting to know whether the treatment of a herniated disc has an influence on the patient's perception of pain. This is done by measuring pain sensation before therapy on a scale of 1 to 50, in the middle of therapy and at the end of therapy. We now want to know if there is a difference between the different moments.

The data are contained in the file `TD06_herniated_disc.txt` in the `data/` folder.

Our question is: **Does the treatment of a herniated disc influence the patient's perception of pain?**

1. Read the data



```r
# Read data file
disc = read.table("data/TD06_herniated_disc.txt", header = TRUE)

# Display start of data file contents
head(disc)
```

2. Data description

```r
# Plot a boxplot representing tablet weight as a function of method
boxplot(pain ~ time, data = disc)
```

**Interpretation of the boxplot**.

It seems that the beginning of the treatment is much more painful than the middle which is slightly more painful than the end of the treatment.

We will now answer the question using Friedman's test as the data are paired.

3. Hypotheses

Specify H0 and H1


<div style = "background-color:#EEE">
H0 : The average pain is the same before, during, and after the therapy
    
H1 : The pain is different at at least one moment during the therapy
</div>

4. Performing the test

```r
# Friedman's test
friedman.test(disc$pain, disc$time, disc$patient)
```

**Conclusion**:

We see p = 0.005 < alpha = 0.05, so we reject H0, which means that the pain the patients experiences along the therapy is different. And we can say, given the previous boxplot, that the therapy decreases the pain.

## 3. Application exercises

### Exercise 1 

A drug is administered to a group of 12 subjects in three modes.
I, taken as a reference, the drug is administered to the subjects on an empty stomach, in mode II
after a normal breakfast and in mode III after a high-fat breakfast. The
results of the blood drug assay, noted for modes of administration I, II and III
respectively y1, y2 and y3 and expressed in ng/L. 

Compare the effect on blood drug levels of modes of administration I, II and III.

The data are contained in the file `TD06_exercise1.txt` in the `data/` folder.

```r
drug = read.table("data/TD06_exercise1.txt", header = TRUE)

head(drug)

# As we're following the same patients across the 3 moments of administration these are paired data
# We can perform a Friedman's test to compare blood drug levels across modes of administration
# with the following hypotheses:
# H0: the drug have the same blood levels across the 3 modes of administrations
# H1: At least one mode of administration show a different drug levels than the others

boxplot(dosing ~ mode, data = drug)

friedman.test(drug$dosing, drug$mode, drug$subject)

# p = 7.76⋅10⁻⁶ <<< alpha = 0.05, so we reject H0, and at least one mode differs from the others.
# Given the boxplot, it seems that mode II give the highest amount of blood drug levels.
```

### Exercise 2:

Three different laboratories A, B, C manufacture ampoules for injectable preparations.
We want to know if the manufacturer's factor has an influence on the ampoule diameters. The
The diameters of 5 ampoules per manufacturer are measured. Data are expressed in mm. 

Answer the following questions: 

1. Is there a significant difference between the three bulbs?
2. If so, test which laboratories obtain different results.

The data are contained in the file `TD06_exercise2.txt` in the `data/` folder.

```r
ampoules = read.table("data/TD06_exercise2.txt", header = TRUE)

# Show data structure
head(ampoules)

# Visualize the data
boxplot(diameter ~ laboratory, data = ampoules)
# => the laboratory seems to produce vastly different diameters ampoules, 
#    but this could be due to small sample size

# Gather sample size
table(ampoules$laboratory)
# => only 5 ampoules per lab, so let's not over-interpret the boxplots

# As the labs are independent, we should perform a Kruskal-Wallis test
# with the following hypotheses:
# H0: ALl the labs produce ampoules of the same diameters
# H1: At least one lab produce ampoules with different diameters

kruskal.test(diameter ~ laboratory, data = ampoules)

# We observe p = 0.024 < alpha = 0.05, so we reject H0 and at least one lab produces ampoules of different diameters
# To test which lab it is we can use pairwise Wilcoxon tests
pairwise.wilcox.test(ampoules$diameter, ampoules$laboratory, p.adjust.method = "none")

# It seems that the B lab produce quite different diameters than the C lab
```
